Question 1: What are two ways to run Python code on your local machine? Which way is preferable for data scientists and why?

````python
# Answer for Question 1
#
# Two common ways to run Python code on a local machine are:
#
#   1. As a script: write the code in a .py file and run it with the Python
#      interpreter from the command line (python my_script.py), or press "Run"
#      in an IDE such as VS Code, PyCharm or Spyder. The whole file is executed
#      top to bottom in a fresh process every time.
#
#   2. Interactively: type code into the Python REPL / IPython shell, or run it
#      cell by cell in a Jupyter notebook (or JupyterLab). The interpreter keeps
#      running between commands, so variables stay in memory.
#
# Jupyter notebooks are usually preferable for data scientists because:
#   - Work is exploratory and iterative: you can load a large dataset once and
#     then try many transformations, models and plots without re-running
#     everything from the start.
#   - Results (tables, charts, images) are shown inline, right next to the code
#     that produced them, which makes it easy to inspect data as you go.
#   - Code, Markdown explanations and outputs live in one document, so the
#     notebook doubles as a report that can be shared with colleagues.
#
# Scripts are still the better choice once an analysis needs to be automated,
# tested, or put into production (e.g. a scheduled data pipeline), because a
# script always runs from a clean state in a fixed order.

print("Preferred for exploratory data science: Jupyter notebooks")
print("Preferred for automation / production: Python scripts (.py files)")
````

Question 2: How do you run a Python script from the command line?

````python
# Answer for Question 2
#
# 1. Open a terminal (Terminal on macOS/Linux, Command Prompt / PowerShell on
#    Windows, or the integrated terminal of your IDE).
# 2. Navigate to the folder that contains the script:   cd path/to/folder
# 3. Call the interpreter with the file name as the argument:
#
#        python my_script.py
#
#    On macOS/Linux the interpreter may be called python3, and on Windows the
#    launcher "py" can also be used (py my_script.py).
#
# Extra arguments can be passed after the file name and are available inside
# the script through sys.argv:
#
#        python my_script.py data.csv 42
#
# Related forms:  python -m module_name   (run an installed module as a script)
#                 python -c "print(1+1)"   (run a short one-off command)

import subprocess, sys, tempfile, os

# Demonstration: write a tiny script to a temporary file and run it exactly as
# you would from the command line.
with tempfile.TemporaryDirectory() as folder:
    script_path = os.path.join(folder, "my_script.py")
    with open(script_path, "w") as f:
        f.write("import sys\nprint('Hello from a script! Arguments:', sys.argv[1:])\n")
    result = subprocess.run([sys.executable, script_path, "data.csv", "42"],
                            capture_output=True, text=True)
    print(result.stdout)
````

Question 3: What is the difference between running Python code in an IDE and a Jupyter notebook?

````python
# Answer for Question 3
#
# IDE (e.g. VS Code, PyCharm, Spyder):
#   - You write .py files and run the whole file from top to bottom in a new
#     Python process each time, so there is no hidden state between runs.
#   - Designed for building software: debugger, linting, auto-completion,
#     refactoring tools, unit-test runners and tight Git integration.
#   - Plain-text files are easy to version-control, review, import as modules
#     and deploy.
#
# Jupyter notebook:
#   - Code is split into cells that are executed one at a time by a persistent
#     "kernel", so variables, loaded data and trained models stay in memory and
#     can be re-used by later cells.
#   - Cells can be run in any order, which is convenient for exploration but can
#     create hidden-state bugs (a cell works only because of something that was
#     run earlier and later deleted).
#   - Rich output (tables, plots, images, HTML) appears inline, and Markdown
#     cells allow narrative text, so a notebook is both analysis and report.
#   - The .ipynb file is JSON that also stores the outputs, which makes it
#     harder to diff/merge in Git and to test automatically.
#
# In short: notebooks are ideal for exploratory analysis, prototyping and
# communicating results; an IDE is ideal for writing reusable, tested,
# production-quality code. Many data scientists use both - prototype in a
# notebook, then move the stable code into .py modules.

print("IDE      -> whole script, fresh process, software-engineering tools")
print("Notebook -> cell by cell, persistent kernel, inline results + Markdown")
````

Question 5: What is the purpose of indentation in Python?

````python
# Answer for Question 5
#
# In Python indentation is part of the syntax, not just a matter of style.
# The indented lines that follow a statement ending in a colon (if, elif, else,
# for, while, def, class, with, try/except ...) form the "block" or "body" that
# belongs to that statement. Other languages use curly braces { } or keywords
# such as begin/end for this; Python uses the indentation itself.
#
#   - All lines in the same block must be indented by the same amount
#     (PEP 8 recommends 4 spaces; never mix tabs and spaces).
#   - Wrong or inconsistent indentation raises an IndentationError, and a line
#     indented differently changes the meaning of the program.
#   - A positive side effect is that Python code is forced to be readable: the
#     visual structure always matches the logical structure.

temperature = 30

if temperature > 25:
    print("It's hot")              # inside the if-block (indented)
    print("Drink some water")      # still inside the if-block
print("This line always runs")     # outside the if-block (not indented)

for i in range(3):
    print("Inside the loop, i =", i)
print("Loop finished")
````

Question 6: What is a variable in Python, and how do you declare one?

````python
# Answer for Question 6
#
# A variable is a name that refers to a value (an object) stored in memory.
# It lets us store a result and refer to it later by name instead of repeating
# the value.
#
# A variable is "declared" simply by assigning to it with the = operator:
#
#       variable_name = value
#
# Unlike languages such as Java or C there is no separate declaration step and
# no type annotation is required - Python is dynamically typed, so the type is
# taken from the value, and the same name can later be bound to a value of a
# different type.
#
# Naming rules: letters, digits and underscores only, cannot start with a
# digit, case-sensitive, and cannot be a reserved keyword (if, for, class ...).
# The convention (PEP 8) is lower-case words separated by underscores.

dataset_name = "titanic"         # a str
number_of_rows = 891             # an int
survival_rate = 0.3838           # a float
is_clean = False                 # a bool

print(dataset_name, number_of_rows, survival_rate, is_clean)

# Several variables can be assigned on one line
x, y, z = 1, 2, 3
print(x, y, z)

# The same name can be re-bound to a value of another type
number_of_rows = "eight hundred and ninety-one"
print(number_of_rows, type(number_of_rows))
````

Question 8: Explain what an object is in Python. What is the relationship between variables and objects?

````python
# Answer for Question 8
#
# In Python *everything* is an object: numbers, strings, lists, functions,
# classes, modules - even True, False and None. An object is a piece of data in
# memory that has
#   - an identity  (a unique id, shown by id(); think of it as its address),
#   - a type       (the class it was created from, shown by type()), which
#                   defines what attributes and methods it has, and
#   - a value      (its contents).
#
# Relationship between variables and objects:
#   A variable is NOT a box that contains a value. It is a name (label) that is
#   bound to an object. Assignment (name = object) attaches the label to the
#   object; it does not copy the object. Consequences:
#     - several variables can refer to the same object (aliasing),
#     - re-assigning a variable just moves the label to a different object,
#       it never changes the object itself,
#     - an object with no labels pointing at it is garbage-collected.

a = [1, 2, 3]
b = a                       # b is now another name for the SAME list object
print(id(a) == id(b), a is b)

b.append(4)                 # change the object through the name b ...
print(a)                    # ... and it is visible through a as well

a = [10, 20]                # re-binding a to a NEW object does not affect b
print(a, b)

# Every object has a type and that type gives it methods
n = 42
print(type(n), type("text"), type([1]), type(print))
print("hello".upper(), (3.75).is_integer())
````

Question 9: What is an operator in Python? List the types of operators that may commonly be used by a data scientist.

````python
# Answer for Question 9
#
# An operator is a symbol (or keyword) that performs an operation on one or more
# values, called operands, and produces a result - e.g. in 3 + 4 the operator
# is + and the operands are 3 and 4.
#
# Types of operator commonly used by a data scientist:
#   - Arithmetic:   +  -  *  /  //  %  **        (calculations, feature engineering)
#   - Comparison:   ==  !=  <  >  <=  >=         (filtering rows, thresholds)
#   - Logical:      and  or  not                 (combining conditions)
#   - Assignment:   =  +=  -=  *=  /=  //=  %=  **=   (storing / updating values)
#   - Membership:   in  not in                   (is a value / column present?)
#   - Identity:     is  is not                   (e.g. x is None)
#   - Bitwise:      &  |  ^  ~  <<  >>           (element-wise boolean masks in
#                                                 NumPy / pandas: df[(a > 1) & (b < 5)])
#   - Matrix multiplication: @                   (linear algebra with NumPy arrays)
#   - Indexing / slicing:  [ ]  and  :           (selecting rows, columns, ranges)

a, b = 17, 5
print("Arithmetic :", a + b, a - b, a * b, a / b, a // b, a % b, a ** 2)
print("Comparison :", a == b, a != b, a > b, a <= b)
print("Logical    :", a > 10 and b < 10, a > 10 or b > 10, not a > 10)

total = 0
total += a                 # augmented assignment
print("Assignment :", total)

columns = ["age", "income", "city"]
print("Membership :", "age" in columns, "salary" not in columns)

missing = None
print("Identity   :", missing is None)
````

Question 11: What is the difference between the `==` operator and the `=` operator?

````python
# Answer for Question 11
#
# =   is the ASSIGNMENT operator. It binds the name on the left to the value on
#     the right:   x = 5   means "let x refer to the value 5". It is a statement
#     and does not produce a value to print or compare.
#
# ==  is the EQUALITY (comparison) operator. It asks whether the two operands
#     have the same value and evaluates to a boolean:  x == 5  is True or False.
#     It does not change anything.
#
# Mixing them up is a classic bug. Python helps here: writing "if x = 5:" is a
# SyntaxError instead of silently assigning, as it would in C.

x = 5                 # assignment: x now refers to 5
print(x == 5)         # comparison: is x equal to 5?  -> True
print(x == 6)         # -> False
print(x)              # x itself is unchanged by the comparisons

# Note: == compares values, "is" compares identity (same object)
print([1, 2] == [1, 2], [1, 2] is [1, 2])
````

Question 19: Explain how Python handles type conversion automatically when using different operators with different types.

````python
# Answer for Question 19
#
# When an operator is applied to two numbers of different types Python performs
# implicit type conversion (also called coercion or type promotion): the
# "narrower" type is converted to the "wider" type so no information is lost.
# The order is   bool -> int -> float -> complex.
#
#   True + 1      -> 2       (bool is treated as int 1)
#   3 + 4.5       -> 7.5     (int converted to float)
#   2 * 3.0       -> 6.0     (int converted to float)
#   1 + 2j + 3    -> (4+2j)  (int converted to complex)
#   10 / 5        -> 2.0     (/ always produces a float)
#
# Python is strongly typed, so it will NOT convert automatically between
# numbers and strings: "3" + 4 raises a TypeError. You must convert explicitly
# with int("3") or str(4). The same operator can also mean different things for
# different types (operator overloading): + adds numbers but concatenates
# strings and lists, and * repeats a string ("ab" * 3 -> "ababab").
#
# Why it matters to a data scientist: data read from CSV files or user input
# often arrives as strings and must be cast before doing arithmetic; and an
# integer column that is combined with a float (or contains a missing value)
# will silently become float, which can change how results are stored and
# displayed.

print(True + 1, type(True + 1))
print(3 + 4.5, type(3 + 4.5))
print(2 * 3.0, type(2 * 3.0))
print(1 + 2j + 3, type(1 + 2j + 3))
print(10 / 5, type(10 / 5))
print(1 == 1.0, True == 1)          # comparisons also convert

print("ab" * 3)                     # * repeats a string
print([1, 2] + [3])                 # + concatenates lists

try:
    print("3" + 4)
except TypeError as error:
    print("TypeError:", error)

print(int("3") + 4, "3" + str(4))   # explicit conversion is required
````

Question 20: Write a Python program to count the number of letters in a given string. Why might a data scientist want to do this?

````python
# Answer for Question 20
text = "Data science combines statistics, programming and domain knowledge."

# len() counts every character - including spaces and punctuation
print("Total characters:", len(text))

# To count only letters, check each character with .isalpha()
letter_count = 0
for character in text:
    if character.isalpha():
        letter_count += 1

print("Letters only:    ", letter_count)

# The same count in a single line
print("Letters only:    ", sum(1 for ch in text if ch.isalpha()))

# Why might a data scientist want to do this?
#   - Text pre-processing / NLP: the length of a message is a simple but useful
#     feature (e.g. spam or sentiment classification, detecting bot reviews).
#   - Data validation and cleaning: checking that fields such as postcodes,
#     IDs or codes have the expected number of characters, or spotting
#     truncated / corrupted entries.
#   - Descriptive statistics on text data: average word length, vocabulary
#     richness, or deciding a maximum sequence length before feeding text to a
#     model.
````

Question 21: What is the difference between mutable and immutable objects in Python? Provide examples. How might this affect how a data scientist develops software in Python?

````python
# Answer for Question 21
#
# Immutable objects cannot be changed after they are created. Any operation that
# appears to modify them actually creates a NEW object and re-binds the name.
#     Examples: int, float, bool, str, tuple, frozenset, bytes, None
#
# Mutable objects CAN be changed in place; the object keeps the same identity.
#     Examples: list, dict, set, bytearray, and most library objects such as
#     NumPy arrays and pandas DataFrames.

# Strings are immutable
s = "data"
print(id(s))
s += " science"            # a brand-new string object is created
print(id(s), s)
try:
    s[0] = "D"
except TypeError as error:
    print("TypeError:", error)

# Lists are mutable
values = [1, 2, 3]
print(id(values))
values.append(4)           # same object, modified in place
values[0] = 100
print(id(values), values)

# Tuples are immutable - a tuple can be used as a dictionary key, a list cannot
coordinates = {(51.5, -0.1): "London"}
print(coordinates[(51.5, -0.1)])

# How does this affect a data scientist?
#   - Aliasing / side effects: passing a list, dict or DataFrame to a function
#     and modifying it there changes the caller's data too. Use .copy() when
#     you need an independent version (a common source of bugs in notebooks,
#     where the same data is shared by many cells).
def add_row(rows):
    rows.append("new row")      # modifies the original list!

original = ["row 1"]
add_row(original)
print(original)

#   - Mutable default arguments (def f(x=[]) ...) keep state between calls.
#   - Only immutable (hashable) objects can be dict keys or set members, which
#     is why tuples are used for composite keys such as (year, month).
#   - pandas / NumPy: some methods return a new object, others modify in place
#     (inplace=True), so you must know which to avoid losing or corrupting data.
#   - Performance: building a large string with += in a loop creates many
#     objects; collect pieces in a list and use "".join() instead.
#   - Immutable objects are safe to share and cache, which makes code easier to
#     reason about and reproduce.
````

Question 22: Write a Python program that creates a list with 10 values in it, then prints the list in reverse order. Why might it be useful to reverse a list of values as a data scientist?

````python
# Answer for Question 22
values = [3, 8, 1, 9, 4, 7, 2, 6, 5, 10]

print("Original list:            ", values)

# 1. Slicing with a step of -1 creates a reversed copy
print("Reversed with slicing:    ", values[::-1])

# 2. The built-in reversed() returns an iterator over the list backwards
print("Reversed with reversed(): ", list(reversed(values)))

# 3. The .reverse() method reverses the list in place (modifies the original)
values_copy = values.copy()
values_copy.reverse()
print("Reversed in place:        ", values_copy)

# Why might reversing a list be useful to a data scientist?
#   - Ordering results: after sorting ascending, reversing gives descending
#     order - e.g. the top-N most important features or highest sales figures.
#   - Time series: data often arrives newest-first (API responses, log files)
#     but must be oldest-first for plotting, rolling averages or model
#     training - or the other way round when showing the latest records.
#   - Back-tracking: replaying steps or events in reverse order (e.g. undoing a
#     sequence of transformations, tracing a path found by an algorithm).
#   - Sequence modelling: some NLP / sequence models process inputs backwards
#     (bidirectional models) and need the reversed sequence as input.
````

Question 23: Write a boolean expression to test whether a temperature variable is greater than or equal to -273 and less than -150. Create within this range to test that the expression works. What might a data scientist be looking for in this range?

````python
# Answer for Question 23
temperature = -200

in_range = temperature >= -273 and temperature < -150
print(f"{temperature} in range: {in_range}")

# Python also allows the comparison to be chained, which reads like maths
in_range = -273 <= temperature < -150
print(f"{temperature} in range: {in_range}")

# Test the expression with values inside, outside and on the boundaries
for temperature in [-300, -273, -200, -151, -150, -100, 20]:
    in_range = temperature >= -273 and temperature < -150
    print(f"{temperature:>5} -> {in_range}")

# What might a data scientist be looking for in this range?
#   - -273 degrees C is (approximately) absolute zero, the lowest temperature
#     that is physically possible. Any reading below it must be a data error
#     (faulty sensor, wrong units, corrupted value), so the lower bound acts as
#     a sanity / data-quality check and the expression can flag impossible
#     readings or confirm a value is physically plausible.
#   - -273 to -150 degrees C is the cryogenic range: liquid nitrogen (-196 C),
#     liquid oxygen (-183 C), superconducting magnets, and temperatures found
#     in deep space or on the outer planets (Neptune is about -200 C). A data
#     scientist could use the expression to select exactly those records -
#     e.g. cryogenic experiments or astronomical observations - from a larger
#     dataset.
````
